```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from xgboost import XGBClassifier

df = pd.read_csv("data/data_NAandSteamOmitted.csv")
```

```python
# Dataset inspection to ensure we've got all the variables we need
print(df.columns.tolist())
# Shape + sample
print(df.shape)
print(df.head())
# What does the data look like per column — types, non-null counts?
print(df.info())
# Quick stats on numeric columns
print(df.describe())

# prep data formats
df['date'] = pd.to_datetime(df['date'])
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
```

```python
# ============================================================
# 2. CREATE TARGET ON FULL DATASET
# ============================================================
df['hvi_category'] = pd.cut(
    df['HVI'],
    bins=[0, 2, 3, 5],
    labels=['Low', 'Medium', 'High']
)

print(f"Dataset size: {len(df):,} rows")
print("\nHVI category distribution:")
print(df['hvi_category'].value_counts())
print("\nSample:")
print(df[['zip', 'date', 'call_count', 'max_temp', 'hvi_category']].head(10))

# ============================================================
# 3. FEATURES + TARGET
# ============================================================
FEATURES = [
    'call_count',         # daily behavioral signal
    'max_temp',           # daily environmental signal
    'tree_count',         # structural — static per ZIP
    'population_density', # structural — static per ZIP
    'population',         # structural — static per ZIP
    'fountains',          # structural — static per ZIP
    'gas_therms'          # structural — proxy for AC usage
]

X = df[FEATURES]
le = LabelEncoder()
y = le.fit_transform(df['hvi_category'])

# ============================================================
# 4. TEMPORAL TRAIN/TEST SPLIT
# Train on 2018-2024, test on 2025 (most recent summer)
# ============================================================
last_year = df['year'].max()
train_mask = df['year'] < last_year
test_mask  = df['year'] == last_year

X_train, X_test = X[train_mask], X[test_mask]
y_train, y_test = y[train_mask], y[test_mask]

print(f"\nTraining on years: {sorted(df[train_mask]['year'].unique())}")
print(f"Testing on year:   {last_year}")
print(f"Train size: {len(X_train):,} | Test size: {len(X_test):,} "
      f"({round(len(X_test)/len(X)*100)}% of data)")

# ============================================================
# 5. MODELS
# Decision Tree: most interpretable, can overfit
# Random Forest: most stable, gives feature importance
# XGBoost: PER JACBO's, typically most accurate
# ============================================================
dt  = DecisionTreeClassifier(max_depth=5, random_state=42)
rf  = RandomForestClassifier(n_estimators=100, random_state=42)
xgb = XGBClassifier(eval_metric='mlogloss', random_state=42)

dt.fit(X_train, y_train);  y_pred_dt  = dt.predict(X_test)
rf.fit(X_train, y_train);  y_pred_rf  = rf.predict(X_test)
xgb.fit(X_train, y_train); y_pred_xgb = xgb.predict(X_test)

# ============================================================
# 6. EVALUATION
# Precision: of all predicted High, how many were actually High?
# Recall: of all truly High ZIPs, how many did we catch?
# F1: balance between precision and recall
# ============================================================
for name, preds in [("Decision Tree", y_pred_dt),
                    ("Random Forest", y_pred_rf),
                    ("XGBoost",       y_pred_xgb)]:
    print(f"\n{name}")
    print(classification_report(y_test, preds, target_names=le.classes_))

# ============================================================
# 7. FEATURE IMPORTANCE (Random Forest)
# ============================================================
importances = pd.Series(rf.feature_importances_, index=FEATURES)
importances.sort_values().plot(kind='barh', color='steelblue')
plt.title("Feature Importances — Full Daily Model (Random Forest)")
plt.xlabel("Importance Score")
plt.tight_layout()
plt.show()

# ============================================================
# 8. COMPARE PREDICTIONS TO ACTUAL HVI
# ============================================================
df['predicted_hvi'] = le.inverse_transform(xgb.predict(X))

print("\nXGBoost Predictions vs Actual HVI Category:")
print(pd.crosstab(df['predicted_hvi'], df['hvi_category'],
                  rownames=['Predicted'], colnames=['Actual']))

# ============================================================
# 9. TOP 5 MOST VULNERABLE ZIPS
# Pulled directly from df — for use in Block 2
# ============================================================
top5_zips = (df.groupby('zip')['HVI']
             .first()
             .nlargest(5)
             .index
             .tolist())

print(f"\nTop 5 most vulnerable ZIPs: {top5_zips}")
```

```python
# ============================================================
# BLOCK 2: TEMPORAL ANALYSIS + Viz
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt

df['date'] = pd.to_datetime(df['date'])
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month

# ============================================================
# 1. IDENTIFY TOP 5 MOST VULNERABLE ZIPS
# Based on highest HVI score in the High category
# ============================================================
top5_zips = (df.groupby('zip')['HVI']
             .first()
             .nlargest(5)
             .index
             .tolist())

print(f"Top 5 most vulnerable ZIPs: {top5_zips}")

df_top5 = df[df['zip'].isin(top5_zips)]

# ============================================================
# 2. VISUALIZATION A — AVERAGE COMPLAINTS BY YEAR
# Shows how behavioral signals shift across summers
# ============================================================
yearly = (df_top5.groupby(['zip', 'year'])['call_count']
          .mean()
          .reset_index())

fig, ax = plt.subplots(figsize=(12, 5))
for zip_code in top5_zips:
    data = yearly[yearly['zip'] == zip_code]
    ax.plot(data['year'], data['call_count'], marker='o', label=str(zip_code))

ax.set_title("Average Daily Complaints by Year — Top 5 Vulnerable ZIPs", fontsize=13)
ax.set_xlabel("Year")
ax.set_ylabel("Avg Daily Complaints")
ax.legend(title="ZIP Code")
plt.tight_layout()
plt.show()

# ============================================================
# 3. VISUALIZATION B — AVERAGE COMPLAINTS BY MONTH
# Shows seasonal pattern within summer months, accross all summers!
# ============================================================
monthly = (df_top5.groupby(['zip', 'month'])['call_count']
           .mean()
           .reset_index())

month_labels = {5: 'May', 6: 'Jun', 7: 'Jul', 8: 'Aug', 9: 'Sep'}
monthly['month_label'] = monthly['month'].map(month_labels)

fig, ax = plt.subplots(figsize=(12, 5))
for zip_code in top5_zips:
    data = monthly[monthly['zip'] == zip_code]
    ax.plot(data['month_label'], data['call_count'], marker='o', label=str(zip_code))

ax.set_title("Average Daily Complaints by Month — Top 5 Vulnerable ZIPs", fontsize=13)
ax.set_xlabel("Month")
ax.set_ylabel("Avg Daily Complaints")
ax.legend(title="ZIP Code")
plt.tight_layout()
plt.show()
```
